1. Criar uma classe qualquer contendo atributos, métodos e construtor


```python
class Jogador:

    def __init__(self, id, nome):
      self.id = id
      self.nome = nome

    def __str__(self):
      return f"ID: {self.id} | Jogador: {self.nome}"

```

2. Criar uma classe herdada desta primeira, ilustrando os conceitos de herança, encapsulamento e polimorfismo

```python
class Classe ( Jogador ):

    def __init__(self, id, nome, classe):
      self._id = None
      self._nome = None
      self._classe = None
      self.classe = classe
      super().__init__(id, nome)

    @property
    def id(self):
      return self._id

    @id.setter
    def id(self, valor):
      if not isinstance(valor, (int)) or valor >= 100:
        print("Valor de id inválido")
      else:
        if self._classe == 'mago':
          self._id = valor + 100
        elif self._classe == 'cavaleiro':
          self._id = valor + 200
        elif self._classe == 'elfo':
          self._id = valor + 300

    @property
    def nome(self):
      return self._nome

    @nome.setter
    def nome(self, valor):
      if not isinstance(valor, str):
        print("Nome inválido")
      else:
          self._nome = valor

    @property
    def classe(self):
      return self._classe

    @classe.setter
    def classe(self, valor):
      if valor not in ['mago', 'cavaleiro', 'elfo']:
        return print("Valor de classe inválido")
      else:
        self._classe = valor

    def __str__(self):
      return f'ID: {self.id} | Jogador: {self.nome} | Classe: {self.classe}'
```

3. Crie instâncias de cada uma das classes, ilustrando o polimorfismo

```python
jogador1 = Jogador(1, 'Pl4yer1')
print(jogador1)
jogador4 = Jogador(4, 1234)
print(jogador4)

classe1 = Classe(1, 'Pl4yer1', 'mago')
print(classe1)
classe2 = Classe(107, 'Gamez12', 'cavaleiro')
classe3 = Classe(56, 'N00b', 'cachorro')
classe4 = Classe(4, 1234, 'elfo')
print(classe4)
```

4. Crie um exemplo para ilustrar a herança múltipla

```python
class TipoConta:

  def __init__(self, tipo):
      self._tipo = tipo
  @property
  def tipo(self):
    return self._tipo

  @tipo.setter
  def tipo(self, valor):
    if valor not in ['ADM', 'DEV', 'USER']:
      return print("Valor de tipo de conta inválido")
    else:
      self._tipo = valor
  def __str__(self):
    return f'Tipo de Conta: {self.tipo}'

class ContaJogo (Jogador, TipoConta):
  def __init__(self, nome, id, tipo):
    Jogador.__init__(self, nome, id)
    TipoConta.__init__(self, tipo)
  def __str__(self):
    return f'CONTA= Jogador: {self.nome} | ID: {self.id} | Tipo de Conta: {self.tipo}'

conta = ContaJogo(99, 'Hero', 'ADM')
print(conta)
```
